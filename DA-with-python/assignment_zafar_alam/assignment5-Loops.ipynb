```python
# Working with Lists
#  Create a list called numbers containing at least 8 integers.
#  Using a for loop, do the following:
#  Print all numbers in the list.
#  Print the sum of all numbers.
#  Print only the even numbers.

numbers = [10, 5, 8, 3, 30, 7, 26, 19]  # create list
total = 0 # define total for sum of all numbers in the list

for number in numbers:
    print(f"Numbers in the list :{number}")  # Print all number from the list through loop
    total = total + number

print(f"\nSum of all numbers: {total} \n")  # Print the sum of all numbers
for number in numbers:
    if number % 2 == 0:
        print(f"Even number in the list: {number}")   # Print even numbers
```

```python
# Dictionary Basics
# Create a dictionary called student with keys (name, age, course, marks (must store a list of 3 marks)
# Using loops: 
# Print all keys and their values.
# Calculate the average of the marks.
# Display whether the student Passed (average ≥ 50) or Failed.
# Requirements:
# Dictionary must contain a list.
# Use loops to calculate the average.

student = {
    "name": "Zafar",
    "age": 20,
    "course": "Python",
    "marks": [75, 60, 80]
}

print("Student Details:") # print heading

for key, value in student.items():
    print(key, ":", value) # Print all keys and their values

total = 0 # for total marks

for mark in student["marks"]:
    total = total + mark

average = total / len(student["marks"]) # Calculate the average.

print(f"\nAverage Marks: {average:.2f}") # print average mark, round on 2 decimal places

if average >= 50:  # Check either the student passed or failed
    print("Result: Passed")
else:
    print("Result: Failed")
```

```python
#  List of Dictionaries
# Create a list called employees.
# Each employee must be stored as a dictionary containing: (name, department, salary)
# Using loops:
# Print details of all employees.
# Find the employee with the highest salary.
# Calculate the total salary expense.
# Requirements:
# 	Minimum 3 employees.
# 	Use comparison logic.
# 	Use loops to process data.

# Create a list of employees
employees = [
    {
        "name": "Ahmed",
        "department": "IT",
        "salary": 50000
    },
    {
        "name": "Zafar",
        "department": "HR",
        "salary": 60000
    },
    {
        "name": "Ali",
        "department": "Finance",
        "salary": 45000
    }
] 
print("Employee Details:") # Print details of all employees

for employee in employees:
    print("Name:", employee["name"])
    print("Department:", employee["department"])
    print("Salary:", employee["salary"])
    print("--------------------")

highest_salary = 0
highest_employee = ""

for employee in employees:
    if employee["salary"] > highest_salary:
        highest_salary = employee["salary"]
        highest_employee = employee["name"]

print("\nEmployee with the highest salary:")
print("Name:", highest_employee)
print("Salary:", highest_salary)

total_salary = 0

for employee in employees:
    total_salary = total_salary + employee["salary"]

print("\nTotal salary expense:", total_salary)
```

```python
# While Loop with User Input
# Ask the user to enter numbers.
# Store the numbers in a list.
# Stop when the user enters -1.
# After the loop ends:
# Print the complete list.
# Print the largest number.
# Print the smallest number.
# Requirements:
# Must use a while loop.# Must store input in a list.
# Handle user input correctly.

numbers = [] # Create an empty list

while True: # Ask the user to enter numbers
    number = int(input("Enter a number (-1 to stop): "))

    if number == -1: # Stop if the user enters -1
        break

    numbers.append(number) # Store the number in the list

if len(numbers) > 0: # Display the results
    print("\nComplete list:", numbers)
    print("Largest number:", max(numbers))
    print("Smallest number:", min(numbers))
else:
    print("\nNo numbers were entered.")
```

```python
# Word Frequency Counter
# Ask the user to enter a sentence.
# Convert the sentence into a list of words.
# Use a dictionary to count how many times each word appears.
# Print the word frequency clearly.
# Example Format (for understanding only):
# apple: 3
# banana: 2
# orange: 1
# Requirements:
# Use string splitting.
# Use a dictionary for counting.
# Use loops for processing.

# Ask the user to enter a sentence
sentence = input("Enter a sentence: ")

# Convert the sentence into a list of words
words = sentence.split()
# print user Input list
print(f"User Input data :{words}")

# Create an empty dictionary
frequency = {}

# Count each word
for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

# Display the word frequency
print("\nWord Frequency:")

for word, count in frequency.items():
    print(word, ":", count)
```

```python
# Nested Loops – Multiplication Table
# Using nested loops, print a multiplication table from 1 to 5.
# Expected Output Format:
# 1 2 3 4 5
# 2 4 6 8 10
# 3 6 9 12 15
# 4 8 12 16 20
# 5 10 15 20 25
# Requirements:
# Must use nested loops.
# Proper formatting required.

# Multiplication table from 1 to 5

for i in range(1, 6):
    row = ""
    for j in range(1, 6):
        row = row + str(i * j) + " " # use STR function to add numeric value
    print(row)
```
