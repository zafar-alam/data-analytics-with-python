```python
# Q4 — Count Frequency of Numbers
# numbers = [1,2,2,3,3,3,4,4,4,4]
# Create frequency dictionary
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

number_frequency = {}

for i in numbers:
    number_frequency[i] = number_frequency.get(i, 0) + 1

print(number_frequency)
```

```python
# Q5 — Word Counter
# Take sentence input and count word frequency.

sentence = input("Enter a complete sentence: ")
print(f"User input : {sentence}")
words = sentence.split()

word_frequency = {}  # inilitize empty dictonary

for i in words:
    if i in word_frequency:
        word_frequency[i] = word_frequency[i] + 1 # count number of sentence occured
    else:
        word_frequency[i] = 1

print("Word frequency:", word_frequency)
```

```python
class Student:

    def __init__(self, name, age, score, math, english, urdu):

        self.name = name
        self.age = age
        self.score = score
        self.math = math
        self.english = english
        self.urdu = urdu

    def get_grade(self, score):

        if score >= 90:
            return "A Grade"
        elif score >= 80:
            return "B Grade"
        else:
            return "C Grade"

    def get_detail(self):

        print("----- Student Detail -----")
        print("Name:", self.name)
        print("Age:", self.age)
        print("Score:", self.score)
        print("Math score:", self.math)
        print("English score:", self.english)
        print("Urdu score:", self.urdu)

        total = self.math + self.english + self.urdu

        print("Total:", total)
        print("Grade:", self.get_grade(total))


student_4 = Student("Zafar", 56, 85, 45, 20, 25)

student_4.get_detail()
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        return f"{self.name} is eating."

    def speak(self):
        return f"{self.name} makes a sound."


class Dog(Animal):
    def speak(self):                    # overrides the parent method
        return f"{self.name} says: Woof!"


dog = Dog("Bruno")
print(dog.eat())      # inherited from Animal
print(dog.speak())    # overridden in Dog
```

```python
class Animal:
    def breathe(self):
        return "Breathing air."

class Mammal(Animal):
    def feed_young(self):
        return "Feeding young with milk."

class Dog(Mammal):
    def speak(self):
        return "Woof!"


dog = Dog()
print(dog.breathe())      # from Animal
print(dog.feed_young())   # from Mammal
print(dog.speak())        # from Dog
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound."


class Dog(Animal):
    def speak(self):
        return f"{self.name} says: Woof!"

class Cat(Animal):
    def speak(self):
        return f"{self.name} says: Meow!"

class Bird(Animal):
    def speak(self):
        return f"{self.name} says: Tweet!"


animals = [Dog("Bruno"), Cat("Whiskers"), Bird("Tweety")]

for animal in animals:
    print(animal.speak())
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def breathe(self):
        return f"{self.name} breathes air."


class Swimmer:
    def swim(self):
        return "Swimming through water."


class Duck(Animal, Swimmer):          # inherits from both
    def speak(self):
        return f"{self.name} says: Quack!"


duck = Duck("Donald")
print(duck.breathe())   # from Animal
print(duck.swim())      # from Swimmer
print(duck.speak())     # Duck's own method
```
