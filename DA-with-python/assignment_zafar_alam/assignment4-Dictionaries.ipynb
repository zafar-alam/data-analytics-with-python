```python
# Q1 — Basic Dictionary Creation
# Create a dictionary called student with keys: name, age, course, city. Print each value.
# student = {
# "name": "",
# "age": ,
# "course": "",
# "city": ""
# }

student = {
 "name": "Zafar",
 "age": 56,
 "course": "Python for Data Analytics",
 "city": "Karachi"
}
print(student)
print("Student's name : " + student["name"])
print("Student's age  : " + str(student["age"]))
print("Course         : " + student["course"])
print("City           : " + student["city"])
```

```python
# Q2 — Adding & Updating Values
# car = {
# "brand": "Toyota",
# "model": "Corolla",
# "year": 2018
# }
# Add color='White', update year to 2022 and print dictionary.

car = {
 "brand": "Toyota",
 "model": "Corolla",
 "year": 2018
 }
print(car)
car["color"] = 'White'

print(car.get('color'))
car['year'] = 2022
print(car)
```

```python
# Q3 — Loop Through Dictionary
# person = {
# "name": "Ali",
# "age": 25,
# "city": "Karachi"
# }
# Print keys and values using loop

person = {
 "Name": "Ali",
 "Age ": 25,
 "City": "Karachi"
 }

for key,value in person.items():
    print(f"{key}  : {value}")

```

```python
# Q4 — Count Frequency of Numbers
# numbers = [1,2,2,3,3,3,4,4,4,4]
# Create frequency dictionary
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

number_frequency = {}

for i in numbers:
    number_frequency[i] = number_frequency.get(i, 0) + 1

print(number_frequency)
```

```python
# Q5 — Word Counter
# Take sentence input and count word frequency.

sentence = input("Enter a complete sentence: ")
print(f"User input : {sentence}")
words = sentence.split()

word_frequency = {}  # inilitize empty dictonary

for i in words:
    if i in word_frequency:
        word_frequency[i] = word_frequency[i] + 1 # count number of sentence occured
    else:
        word_frequency[i] = 1

print("Word frequency:", word_frequency)
```

```python
# Q6 — Student Marks & Average
# marks = {"Math":80, "English":70, "Physics":90}
# Calculate total and average.

marks = {"Math":80, "English":70, "Physics":90}

total_marks = 0
average_marks = 0

total_marks = marks["Math"] + marks["English"] + marks["Physics"]
average_marks = total_marks / len(marks) # len function will calculate number of eliments in marks dictionary.

print(f'Total subjects : {len(marks)}')
print(f"Total Marks    : {total_marks}")
print(f"Average Marks  : {average_marks}")
```

```python
# Q7. Highest Scoring Student
### students = {
# "Ali":85,
# "Sara":92,
# "Ahmed":78,
# "Zara":95
# }
#Find highest scoring student """

students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}
student_name = max(students, key=students.get) # through max function find the key first then throguh key get the its value.
student_score = students[student_name]

print(f"Student's name is {student_name} and their highest score is {student_score}")
```

```python
# Q8 — Merge Dictionaries
# dict1 = {"a":1,"b":2}
# dict2 = {"c":3,"d":4}
# Merge and print result

dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged = {}  # initlize empty new dictonary

merged["a"] = dict1["a"]
merged["b"] = dict1["b"]
merged["c"] = dict2["c"]
merged["d"] = dict2["d"]

print(merged)
```
