```python
# Write a program that rotates the element of a list so that the element at the first index moves to the
# second index, the element in the second index moves to the third index, etc., and the element in the last
# index moves to the first index. 

myList = [25,10,15,95,20]
LastElement = [myList[-1]]      # pick last item from list [20] 
WithoutLastElement = myList[:-1] # pick list excluding last item [25,10,15,95]

rotatedList = LastElement + WithoutLastElement  # concate both lists in new list: [20] + [25, 10, 15, 95]

print("Original list:", myList)
print("Rotated list:", rotatedList)
```

```python
# Write a program to multiply two matrices 
matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]

matrix2 = [
    [7, 8],
    [9, 10],
    [11, 12]
]

result = [
    [matrix1[0][0]*matrix2[0][0] + matrix1[0][1]*matrix2[1][0] + matrix1[0][2]*matrix2[2][0],
     matrix1[0][0]*matrix2[0][1] + matrix1[0][1]*matrix2[1][1] + matrix1[0][2]*matrix2[2][1]],

    [matrix1[1][0]*matrix2[0][0] + matrix1[1][1]*matrix2[1][0] + matrix1[1][2]*matrix2[2][0],
     matrix1[1][0]*matrix2[0][1] + matrix1[1][1]*matrix2[1][1] + matrix1[1][2]*matrix2[2][1]]
]

print("Matrix 1:", matrix1)
print("Matrix 2:", matrix2)
print("Product:", result)
```
