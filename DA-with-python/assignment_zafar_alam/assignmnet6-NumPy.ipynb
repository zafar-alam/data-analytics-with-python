```python
# Create a 1D NumPy array of integers from 10 to 50 (inclusive, step 5). Print the array, its shape, dtype and number of dimensions.

import numpy as np

my_array = np.arange(10, 51, 5)

# Print the array and its properties
print("Array:", my_array) # Print eliments of the array
print("Array Shape:", my_array.shape) # Print Number of eliments in the array 
print("Array Data type:", my_array.dtype)
print("Number of dimension:", my_array.ndim)
```

```python
# Create a 3x4 array of zeros, a 3x4 array of ones, and a 4x4 identity matrix. 
# Then create an array of 8 evenly spaced numbers between 0 and 1 using linspace.

zeros_array = np.zeros((3, 4))
print("Array of Zeros:\n", zeros_array)

ones_array = np.ones((3, 4))
print("\nArray of Ones:\n", ones_array)

identity_array = np.identity(4)  # There is another method eye(), used for create Identity maxtrix
print("\nIdentity Matrix:\n", identity_array)

linspace_array = np.linspace(0, 1, 8)
print("\n8 Evenly spaced numbers:")
print(linspace_array)
```

```python
# Create a 1D array of numbers from 1 to 24 and reshape it into a 2D array of shape (4, 6). 
# Then reshape it into a 3D array of shape (2, 3, 4) and print the shape of each.

arr_1d = np.arange(1, 25) # Create a 1D array from 1 to 24
arr_2d = arr_1d.reshape(4, 6) # Reshape the 1D array into a 2D array of shape (4, 6)
arr_3d = arr_1d.reshape(2, 3, 4) # then Reshape 1D array into a 3D array of shape (2, 3, 4)

print("1D Array Shape:", arr_1d.shape)
print("2D Array Shape:", arr_2d.shape)
print("3D Array Shape:", arr_3d.shape)
```

```python
# Given arr = np.array([12, 45, 7, 23, 56, 89, 34, 21]), 
# print: the first 3 elements, the last 3 elements, every alternate element, and the array in reverse order.

my_arr = np.array([12, 45, 7, 23, 56, 89, 34, 21])

print("First 3 elements:", my_arr[:3]) # The first 3 elements through slicing
print("Last 3 elements:", my_arr[-3:]) # The last 3 elements through slicing
print("Every alternate element:", my_arr[::2]) # Every alternate element trhough slicing
print("Array in reverse order:", my_arr[::-1]) # The array in reverse order (starting from end)
```

```python
# Create a 2D array of shape (5, 5) with values 1 to 25. 
# Extract (a) the second row, (b) the third column, (c) the 3x3 sub-matrix from the center.

# 1. Create a 2D array of shape (5, 5) with values from 1 to 25
arr = np.arange(1, 26).reshape(5, 5)
print("-" * 30)
print("Original 5x5 Array:\n", arr)

second_row = arr[1] # Get only second row with all column through 1 index
print("\nSecond row:\n", second_row)

third_column = arr[:, 2] # Get only 3 Column with all row 
print("\nThird column:\n", third_column)

# (c) 
center_submatrix = arr[1:4, 1:4]  # Extract the 3x3 sub-matrix from the center of Rows 1 to 3 (stop at 4) and Columns 1 to 3 (stop at 4)
print("\n3x3 Center Sub-matrix:\n", center_submatrix)
print("-" * 30)
```

```python
# Given a = np.array([10, 20, 30, 40]) and b = np.array([1, 2, 3, 4]), 
# perform addition, subtraction, multiplication, division and power (a ** 2). Also compute the dot product of a and b.
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Addition (a + b):", a + b)
print("Subtraction (a - b):", a - b)
print("Multiplication (a * b):", a * b)
print("Division (a / b):", a / b)
print("Power (a ** 2):", a ** 2)
print("Dot Product (np.dot(a, b)):", np.dot(a, b))
```

```python
# Given arr = np.array([[4, 8, 2], [9, 1, 7], [5, 6, 3]]), find the sum, mean, min, max and standard deviation of the whole array, 
# and then the sum along rows (axis=1) and along columns (axis=0).
arr = np.array([[4, 8, 2], [9, 1, 7], [5, 6, 3]])

print("--- Overall Statistics ---")
print("Sum:", arr.sum())                  # or np.sum(arr)
print("Mean:", arr.mean())                # or np.mean(arr)
print("Min:", arr.min())                  # or np.min(arr)
print("Max:", arr.max())                  # or np.max(arr)
print("Standard Deviation:", arr.std())    # or np.std(arr)
print("\n--- Axis Operations ---")
print("Sum along rows (axis=1):", arr.sum(axis=1))
print("Sum along columns (axis=0):", arr.sum(axis=0))
```

```python
# Create an array of 20 random integers between 1 and 100. Using boolean masking, 
# print (a) all elements greater than 50, (b) all even elements, (c) count of elements divisible by 5.

arr = np.random.randint(1, 101, size=20)
print("Original Array:\n", arr)
print("-" * 40)
gt_50 = arr[arr > 50]
print("(a) Elements greater than 50:", gt_50)

even_elements = arr[arr % 2 == 0]
print("(b) Even elements:", even_elements)

div_by_5_count = np.sum(arr % 5 == 0)
print("(c) Count of elements divisible by 5:", div_by_5_count)
```

```python
# Use np.where to replace all negative values in np.array([5, -3, 8, -1, 0, -7, 4]) with 0. 
# Also find the index of the maximum and minimum value using argmax and argmin.

arr = np.array([5, -3, 8, -1, 0, -7, 4])
print("Original Array:", arr)
print("-" * 40)

replaced_arr = np.where(arr < 0, 0, arr) # get help through chatgpt
print("Array after replacing negatives with 0:", replaced_arr)

max_index = np.argmax(arr)
print("Index of maximum value (8):", max_index)

min_index = np.argmin(arr)
print("Index of minimum value (-7):", min_index)
```

```python
# Create two 2x3 arrays and join them vertically (vstack) and horizontally (hstack). 
# Then sort the array np.array([[9, 3, 6], [2, 8, 1]]) along each row and flatten the result into 1D.
arr1 = np.array([[1, 2, 3], [4, 5, 6]])

arr2 = np.array([[7, 8, 9], [10, 11, 12]])

v_stacked = np.vstack((arr1, arr2))
print("Vertical Stack (vstack):\n", v_stacked)

h_stacked = np.hstack((arr1, arr2))
print("\nHorizontal Stack (hstack):\n", h_stacked)
print("-" * 40)

arr_to_sort = np.array([[9, 3, 6], [2, 8, 1]])

sorted_rows = np.sort(arr_to_sort, axis=1)
print("Sorted along rows:\n", sorted_rows)

flattened = sorted_rows.flatten()
print("Flattened 1D array:", flattened)
```
