### Write a program in which use all the operators we can use in Python

```python
## I'm using \n escape sequance for add new line
a = 7
b = 3

# 1. Arithmetic Operators
print("Arithmetic Operators")
print ("-------------------")
print("Add:", a + b)
print("Subtract:", a - b)
print("Multiply:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exp:", a ** b)
print("\n")
# 2. Comparison Operators
print("Comparison Operators")
print ("-------------------")
print("Equal:", a == b)
print("Not Equal:", a != b)
print("Greater Than:", a > b)
print("Less Than:", a < b)
print("Greater Than or Equal:", a >= b)
print("Less Than or Equal:", a <= b)
print("\n")
# 3. Assignment Operators
print("Assignment Operators")
print ("-------------------")
z = 10
z += 2
print("z += 2:", z)
z -= 2
print("z -= 2:", z)
z *= 2
print("z *= 2:", z)
z /= 2
print("z /= 2:", z)
print("\n")
# 4. Logical Operators
print("Logical Operators")
print ("----------------")
print("AND:", a > 5 and b < 5)
print("OR:", a > 15 or b < 5)
print("NOT:", not(a > 5))
```

# Completes the following steps of small task:
# Mention Marks of English , Islamiat and Maths out of 100 in 3 different variables
# Mention Variable of Total Marks and assign 300 to it
# Calculate Percentage

```python
marks_eng = int(input("Please enter English marks :"))
marks_islam = int(input("Please enter Islamyat marks :"))
marks_math = int(input("Please enter Maths marks :"))
total_marks = 300
obtain_marks = marks_eng + marks_islam + marks_math
percentage = (obtain_marks / total_marks ) * 100
print("\n")
print("Marks Sheet ")
print("-------------")
print("English marks :\t\t" + str(marks_eng))
print("Islamyat marks :\t" + str(marks_islam))
print("Maths marks :\t\t" + str(marks_math))
print("\n")
print("Total marks :\t\t" + str(total_marks))
print("Obtain marks :\t\t" + str(obtain_marks))
print("Your percentage % :\t" + str(percentage))
```

### Take the values of two variables a and b, then swap their values without using a third variable and print the result before and after swapping

```python
a = int(input("Enter value of a: "))
b = int(input("Enter value of b: "))

print("Before swapping:")
print("a =", a)
print("b =", b)

# Swapping without a third variable
a = a + b
b = a - b
a = a - b

print("\nAfter swapping:")
print("a =", a)
print("b =", b)
```

### Take the radius of a circle from a variable and calculate its Area and Circumference (use pi = 3.14159)

```python
pi = 3.14159

radius = int(input("Enter the radius of the circle: "))

area = pi * radius * radius
circumference = 2 * pi * radius

print("Area of circle:", area)
print("Circumference of circle:", circumference)
```

### Take the price of an item and a discount percentage in variables, then calculate and print the discount amount and the final price after discount

```python
item_price = int(input("Please enter Item price"))
discount = int(input("Please enter discount percentage"))
disc_amt = (item_price/discount)
print("The amount before discount :\t" + str(item_price))
print("The discount amount on the item :" + str(disc_amt)) 
print("The total amount after discount :" +str(item_price - disc_amt))
```
