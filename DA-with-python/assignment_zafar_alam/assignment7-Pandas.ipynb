```python
# From the DataFrame of Q1, select (a) only the Name and Salary columns, (b) rows where Age is greater than 30, (c) rows where City is 'Delhi' and Salary is above 40000.

data = {
    'Name': ['Ali', 'Zafar', 'Ahmed', 'Ayan', 'Baber', 'Sarim'],
    'Age': [28, 56, 45, 29, 38, 42],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Delhi', 'Sydney'],
    'Salary': [70000, 35000, 120000, 75000, 41000, 110000]
}

df = pd.DataFrame(data)

# (a) Select only the Name and Salary columns
print("(a) Name and Salary columns:\n", df[['Name', 'Salary']])
print("-" * 40)

# (b) Filter rows where Age is greater than 30
print("(b) Rows where Age > 30:\n", df[df['Age'] > 30])
print("-" * 40)

# (c) Filter rows where City is 'Delhi' and Salary is above 40000
print("(c) Rows where City is 'Delhi' and Salary > 40000:\n", df[(df['City'] == 'Delhi') & (df['Salary'] > 40000)])
```

```python
# Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. 
# Then rename the column 'City' to 'Location' and drop the 'Bonus' column.
df = pd.DataFrame(data)

# Add a new column 'Bonus' that is 10% of Salary
df['Bonus'] = df['Salary'] * 0.10
print(df)
print("-" * 60)

# Add a column 'Total' = Salary + Bonus
df['Total'] = df['Salary'] + df['Bonus']
print(df)
print("-" * 60)

# Rename the column 'City' to 'Location'
df = df.rename(columns={'City': 'Location'})
print(df)
print("-" * 60)

# Drop the 'Bonus' column
df = df.drop(columns=['Bonus'])

# Display the final modified DataFrame
print(df)
```

```python
# Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, 
# fill numeric NaNs with the column mean, and drop rows that still have missing values.
data = {
    'Name': ['Ali', 'Zafar', 'Ahmed', 'Ayan', 'Baber'],
    'Age': [28, None, 45, 29, 38],
    'Salary': [70000, 85000, 120000, 95000,None],
    'Rating': [4.2, None, None, 3.9, 4.5]
}

df = pd.DataFrame(data)
print("--- Original DataFrame ---")
print(df)
print("-" * 40)

# 1. Find the count of missing values in each column
missing_counts = df.isnull().sum()
print("Count of missing values per column:\n", missing_counts)
print("-" * 40)

# 2. Fill numeric missing values with the column mean
# We use df.select_dtypes(include=['number']) instead of numpy
numeric_cols = df.select_dtypes(include=['number']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

print("DataFrame after filling numeric missing values with column means:\n", df)
print("-" * 40)

# 3. Drop rows that still have missing values (e.g., the None in the 'Name' column)
df_cleaned = df.dropna()
print("DataFrame after dropping remaining rows with missing values:\n", df_cleaned)
```

```python
# Sort the DataFrame by Salary in descending order, and then by Age in ascending order. 
# Also print the top 3 highest-paid employees using nlargest().
df = pd.DataFrame(data)

# 1. Sort by Salary in descending order, and then by Age in ascending order
sorted_df = df.sort_values(by=['Salary', 'Age'], ascending=[False, True])
print("Sorted DataFrame (Salary DESC, Age ASC) ---")
print(sorted_df)
print("-" * 50)

# 2. Print the top 3 highest-paid employees using nlargest()
top_3_paid = df.nlargest(3, 'Salary')
print("Top 3 Highest-Paid Employees ---")
print(top_3_paid)
```

```python
# Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. 
# Show the result using agg() with multiple functions.

data = {
    'Name': ['Ali', 'Zafar', 'Ahmed', 'Ayan', 'Baber', 'Sarim'],
    'Age': [28, 56, 45, 29, 38, 42],
    'City': ['New York', 'Delhi', 'Paris', 'Tokyo', 'Delhi', 'Sydney'],
    'Salary': [70000, 35000, 120000, 75000, 41000, 110000]
}

df = pd.DataFrame(data)

# Group by 'City' and apply multiple functions to the 'Salary' column
city_stats = df.groupby('City')['Salary'].agg(['mean', 'count', 'max'])

print("--- Salary Statistics per City ---")
print(city_stats)
```

```python
# Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, 
# then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.
import pandas as pd

# 1. Create employees DataFrame
df_emp = pd.DataFrame({
    'EmpID': [1, 2, 3, 4, 5],
    'Name': ['Ali', 'Zafar', 'Ahmed', 'Ayan', 'Baber'],
    'DeptID': [101, 102, 101, 103, 105]  # DeptID 105 has no matching department
})

# 2. Create departments DataFrame
df_dept = pd.DataFrame({
    'DeptID': [101, 102, 103, 104],
    'DeptName': ['HR', 'IT', 'Finance', 'Marketing']  # DeptID 104 has no employees
})

# --- MERGING ---

# Inner Join: Keeps only rows where DeptID matches in both DataFrames
inner_merged = pd.merge(df_emp, df_dept, on='DeptID', how='inner')
print("--- Inner Join ---\n", inner_merged)
print("-" * 50)

# Left Join: Keeps all rows from df_emp, adding NaN where DeptID doesn't match in df_dept
left_merged = pd.merge(df_emp, df_dept, on='DeptID', how='left')
print("--- Left Join ---\n", left_merged)
print("-" * 50)

# --- CONCATENATION & DEDUPLICATION ---

# Create two DataFrames with the same structure (including an intentional duplicate row for 'Ahmed')
df_part1 = pd.DataFrame({
    'Name': ['Ali', 'Zafar', 'Ahmed'],
    'Salary': [70000, 85000, 120000]
})

df_part2 = pd.DataFrame({
    'Name': ['Ahmed', 'Ayan', 'Baber'],
    'Salary': [120000, 75000, 95000]
})

# Concatenate vertically and remove duplicate rows
concat_df = pd.concat([df_part1, df_part2]).drop_duplicates().reset_index(drop=True)
print("--- Concatenated & Deduplicated DataFrame ---\n", concat_df)
```
