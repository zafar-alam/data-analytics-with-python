## Assignment-8  Querying, Sorting & Filtering

```python
use bikestores;

-- Question 1 ---------------------------
select first_name, last_name, city, phone,state 
	from sales.customers
	where state = 'CA' and phone is not null;

-----------------------------------------

-- Question 2 ---------------------------
select p.product_id, p.product_name, p.model_year,p.list_price 
	from production.products p
	order by p.model_year desc, p.list_price asc;

-----------------------------------------

-- Question 3 ---------------------------
----  Part A ------
select top 5 p.product_name, p.list_price 
	from production.products p
	order by p.list_price desc;
----  Part B ------
select top 5 PERCENT *
	from production.products
	order by list_price asc;
-- Row count: 5% of the total number of rows in Products, 
-- rounded up to a whole number.

-----------------------------------------

-- Question 4 ---------------------------
---   a) Page 1  (rows  1 – 10)
select *
	from production.products
	order by list_price desc 
	offset 0 rows
	fetch next 10 rows only;
-----------------------------------------
--- b) Page 2  (rows 11 – 20)
select *
	from production.products
	order by list_price desc 
	offset 10 rows
	fetch next 10 rows only;
-----------------------------------------
--- c) Page 3  (rows 21 – 30)
select *
	from production.products
	order by list_price desc 
	offset 20 rows
	fetch next 10 rows only;

-----------------------------------------

-- Question 5 ---------------------------
--- Part A ----
select distinct c.state
	from sales.customers c
	order by c.state asc ;
--- Part B ----
select distinct c.state, c.city
	from sales.customers c
	order by c.state asc, c.city asc ;
--- Part C ----
select  DISTINCT model_year   	
	from production.products;
--- total 4 Records count
--- with count function --
select  count(DISTINCT model_year) unique_model_years   	
	from production.products;

-----------------------------------------
---  Question 6 — Logical Operators (AND / OR)
SELECT product_id, product_name, brand_id, category_id, list_price
	FROM production.products
	WHERE list_price BETWEEN 500 AND 1500
  		--AND (model_year = 2019 OR model_year = 2020)
	ORDER BY list_price ASC;

-----------------------------------------	
```
