```python
# Pie chart
# used to show part-to-whole relationships/data

import matplotlib.pyplot as plt

# data
categories = ['A', 'B', 'C', 'D', 'E']
sales = [10, 20, 55, 35, 45]

# 1. Create custom legend labels combining category name and value
legend_labels = [] # create empty list

for cat, val in zip(categories, sales):  # Start loop for get values from categories and sales lists
    label = f'{cat}: {val}'
    legend_labels.append(label)  # Insert the label values like A:10, B:20 and so on in legend_labels list

# legend_labels = [f'{cat}: {val}' for cat, val in zip(categories, sales)]

# 2. Plot the pie chart (omitting 'labels=categories' to keep the chart clean)
wedges, texts, autotexts = plt.pie(
    sales, 
    labels=categories,
    autopct='%1.1f%%', 
    startangle=90
)

# 3. Add the legend with the custom labels and position it outside the chart
plt.legend(
    wedges, 
    legend_labels, 
    title="Categories & Values", 
    loc = "center left", 
    bbox_to_anchor=(1, 0, 0.5, 1)
)

plt.title("Pie Chart Example")

# Adjust layout so the legend isn't cut off
plt.tight_layout()
plt.show()
```

```python
# Sub-Plots in chart
# data-1 - bar chart for Week 1
categories = ['Mon','Tue','Wed','Thu','Fri']
sales = [10, 20, 55, 35, 45]
plt.figure(figsize=(15,5))


# first plot- bar chart
plt.subplot(1,3,1) # row, column, position
plt.bar(categories, sales)
plt.grid()
plt.title("Daily Sales Week-1")
plt.xlabel("Week Days")
plt.ylabel("Sales")

# data-2 - bar chart for Week 2
categories = ['Mon','Tue','Wed','Thu','Fri']
sales = [20, 15, 35, 45, 30]

# second plot- Pie  chart with 2nd week sales
plt.subplot(1,3,2) # row, column, position
#plt.bar(categories, sales)
plt.pie(sales, labels=categories, autopct = '%1.1f%%', startangle=90)
plt.title("Daily Sales Week-2")
plt.xlabel("Week Days")
plt.ylabel("Sales")

# Third Plot scatter chart with 2nd week sales
plt.subplot(1,3,3) # row, column, position
plt.scatter(categories, sales)
plt.title("Daily Sales Week-2")
plt.xlabel("Week Days")
plt.ylabel("Sales")

plt.savefig("Monthly_Sales.png") # Save the chart into image file


plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define a layout: 'A' spans the top row, 'B' and 'C' share the bottom row
mosaic_layout = """
    AAA
    BC.
    """

fig, axs = plt.subplot_mosaic(mosaic_layout, figsize=(8, 5), layout="constrained")

# Access subplots directly by their string labels
axs['A'].plot(np.sin(np.linspace(0, 10, 100)), color='royalblue')
axs['A'].set_title("Top Panel (Spans 3 Columns)")

axs['B'].bar(['X', 'Y', 'Z'], [5, 7, 3], color='coral')
axs['B'].set_title("Bottom Left")

axs['C'].scatter(np.random.rand(10), np.random.rand(10), color='forestgreen')
axs['C'].set_title("Bottom Center")

plt.show()
```
