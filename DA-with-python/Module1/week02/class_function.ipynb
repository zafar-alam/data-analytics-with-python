```python
def get_grade(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    else:
        return "Fail"


students = [
    {"name": "Ali", "score": 92},
    {"name": "Sara", "score": 75},
    {"name": "Ahmed", "score": 55},
    {"name": "Zara", "score": 88}
]

for student in students:
    grade = get_grade(student["score"])
    print(student["name"], "-", grade)
```

```python
# Calculate Tax
# Bill amount
# tax percentage 15%
# calcuate bill amount with Tax
# return bill amount

def calculate_tax(bill_amount):
    tax = bill_amount * 15 / 100
    total_bill = bill_amount + tax
    discount = bill_amount * 10 / 100
    net_bill = total_bill - discount 
    return net_bill


bill = 1000

final_bill = calculate_tax(bill)

print("Bill amount:", bill)
print("Bill with tax:", final_bill)
print("Bill discount:", discount)
```
